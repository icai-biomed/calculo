# Newton's Method for $\sqrt{2}$

```python
def f(x):
  return(x**2 - 2)
```

```python
def df(x):
  return(2 * x)
```

```python
def newton_step(x):
  return(x - f(x) / df(x))
```

```python
newton_step(1.4)
```

```python
newton_step(newton_step(1.4))
```

```python
newton_step(newton_step(newton_step(newton_step(1.4))))
```

```python
import numpy as np
np.sqrt(2)
```

```python
x0 = 1.4
print(f"Root candidate {x0}, function value {f(x0)} ")
k = 5
for i in np.arange(k):
  x0 = newton_step(x0)
  print(f"Root candidate {x0}, function value: {f(x0)} ")
```

### Newton's Method using SymPy for Function and Derivative

```python
# Define a symbolic variable
x_sym = sy.Symbol('x')

# Define the symbolic function: sin(x)^2 - 1/2
f_sym = sy.sin(x_sym/4)**2 - sy.S(1)/2

# Compute the symbolic derivative
df_sym = sy.diff(f_sym, x_sym)

print(f"Symbolic function: {f_sym}")
print(f"Symbolic derivative: {df_sym}")
```

```python
# Convert symbolic functions to numerical functions using lambdify
f_numeric = sy.lambdify(x_sym, f_sym, 'numpy')
df_numeric = sy.lambdify(x_sym, df_sym, 'numpy')

# Define the newton_step using the numeric functions from sympy
def newton_step_sympy(x_val):
  return x_val - f_numeric(x_val) / df_numeric(x_val)

# Perform the Newton's method computation
x0_sympy = 1.4
k_sympy = 15

print(f"Starting Newton's method with x0 = {x0_sympy}, k = {k_sympy} iterations (using sympy derived functions)\n")
print(f"Root candidate {x0_sympy}, function value {f_numeric(x0_sympy)}")
for i in np.arange(k_sympy):
  x0_sympy = newton_step_sympy(x0_sympy)
  print(f"Root candidate {x0_sympy}, function value: {f_numeric(x0_sympy)}")
```
