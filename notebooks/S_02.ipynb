# Approximate slope of the tangent line

# Tangent line and approximate values

 We use sympy to define $f(x) = \sin(x)$ and fix the point $x_0$ where the tangent line will be computed

```python
x = sy.Symbol('x', real=True)
f = sy.sin(x)

x0 = 0
```

We define a function called approx_slope that takes $x_0$ and a very close point $x_1$ and computes the approximate value of the slope of f at $x_0$

```python
def approx_slope(f, x0, x1):
  m = (f.subs(x, x1) - f.subs(x, x0)) / (x1 - x0)
  return m
```

Now let us find an approximate value of the slope at x1 = 0.001

```python
x1 = 0.00001
approx_slope(f, x0, x1)
```

What do you think of these values? What is the tangent line to $f$ at $x_0$?

### Tangent line to $f(x) = \sin(x)$ at $x_0 = 0$

```python
df = sy.diff(f, x)
print(f"f(x) = {f}")
print(f"f'(x) = {df}")
```

Now we will use the equation of the tangent line at $x_0 = 0$:
$$y = f(x_0) + f'(x_0) (x - x_0)$$
That is:
$$y = \sin(0) + \cos(0) (x - 0) = 0 + 1\cdot x = x$$
The tangent line is therefore $y = x$

```python
# We obtain the tangent line in sympy
# To do that we evaluate the function and its derivative at 0
f_x0 = f.subs(x, x0)
df_x0 = df.subs(x, x0)
# And substitute those values in the equation of the tangent line
tl = f_x0 + df_x0 * (x - x0)
tl
```

### Approximate values using the tangent line

Now take $x1 = 0.003$. The approximate value provided by the tangent line is $tl(x1)$, which in the sin function at the origin example is just $x1$.

```python
# Let us define x1 = 0.003 in sympy
x1 = sy.Float(0.003)

def apprx_value(f, x0, x1):
  f_x0 = f.subs(x, x0)
  df_x0 = df.subs(x, x0)
  tl = f_x0 + df_x0 * (x - x0)
  return tl.subs(x, x1)

apprx_value(f, x0, x1)
```

Let us now ask sympy for a direct estimate for comparison.

```python
# sympy value of f at x1
f.subs(x, x1)
```

Plotting functions is usually much faster and flexible using tools such as GeoGebra or Wolfram Alpha. But ocasionally we will need to make plots in python. For a preliminary example:

```python
sy.plot(f, (x, -sy.pi, sy.pi), title='Plot of f(x) = sin(x)');
```

Convert a sympy (symbolic) function to a (numeric) numpy-usable python function:

```python
f_numpy = sy.lambdify(x, f, 'numpy')
```

And now plot it with MatplotLib, a very popular plotting python library. MatplotLLib plots can be very elaborate, and a bit intimidating at first. Don't worry, you will soon get used to it.

```python
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(-np.pi, np.pi, 100)
y_vals = f_numpy(x_vals)

plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Plot of f_numpy(x) = sin(x) using Matplotlib')
plt.grid(False) # Enabled grid
plt.axhline(0, color='black',linewidth=0.5) # Add x-axis line
plt.axvline(0, color='black',linewidth=0.5) # Add y-axis line
plt.show()
```
