## Removable discontinuity

 We use sympy to define
 $$f(x) = \begin{cases}
\dfrac{\sin(x)}{x} & \text{ si }x\neq 0\\[2mm]
\quad 1 & \text{ si }x =  0
\end{cases}$$
and fix the point $x_0$ where the tangent line will be computed

```python
x = sy.Symbol('x', real=True)
```

```python
# Define the piecewise function
f = sy.Piecewise((sy.sin(x)/x, sy.Ne(x, 0)), (1, sy.Eq(x, 0)))

print("f(x) =")
display(f)
```

And we plot the function:

```python
sy.plot(f, (x, -10, 10), title='Plot of f(x)');
```

```python
f_numpy = sy.lambdify(x, f, 'numpy')

import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(-10, 10, 100)
y_vals = f_numpy(x_vals)

plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Plot of f_numpy(x) = sin(x)/x using Matplotlib')
plt.grid(False) # Enabled grid
plt.axhline(0, color='black',linewidth=0.5) # Add x-axis line
plt.axvline(0, color='black',linewidth=0.5) # Add y-axis line
plt.show()
```

## Computing limits with sympy

Let us use sympy to compute the limit of the function $f$ we have used above.

```python
sy.limit(f, x, 0)
```

## Computing indeterminate form limits with sympy

Define the expression

```python

expr = (sy.tan(x) - sy.sin(x)) / sy.log(1 + x**3)

print("Expression:")
display(expr)
```

And now simply compute the limit as x -> 0

```python
limit_val = sy.limit(expr, x, 0)

print("\nLimit as x -> 0:")
display(limit_val)
```

## Computing one-sided (lateral) limits

We begin by defining a function:

```python
f = 1 / (1 + sy.exp(1/x))

print("Function f(x) =")
display(f)
```

Now we compute a one-sided limit (the left-hand one) as follows. Note the use of `dir`:

```python
left_limit = sy.limit(f, x, 0, dir='-')

print("\nLeft-hand limit of h(x) as x -> 0:")
display(left_limit)
```

Similarly for the right hand limit:

```python
right_limit = sy.limit(f, x, 0, dir='+')

print("\nRight-hand limit of h(x) as x -> 0:")
display(right_limit)
```

And a plot that confirms visually these results:

```python
# Plot f
sy.plot(f, (x, -10, 10), title='Plot of f(x)');
```
