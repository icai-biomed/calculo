## Infinite limit


 We use sympy to define
 $$f(x) = \dfrac{x^4 + 3}{x^2 - x -2}$$

```python
x = sy.Symbol('x', real=True)
```

```python

f = (x**4 + 3) / (x**2 - x - 2)

print("f(x) =")
display(f)
```

And we plot the function:

```python
sy.plot(f, (x, -10, 10), ylim=(-50, 50), title='Plot of f(x)');
```

Next we convert it to numpy and we use `linspace` to get negative powers of 10 and use them to define a sequence of values like 2.1, 2.01, 2.001, etc. that get closer and closer to 2.  
Note that these values are in decreasing order.

```python
f_numpy = sy.lambdify(x, f, 'numpy')

import numpy as np
import matplotlib.pyplot as plt

x_vals = 2 + 10 ** np.linspace(-1, -10, 10)
print(f"The first five x_vals are {[f'{val:.10f}' for val in x_vals[:5]]}")
y_vals = f_numpy(x_vals)
```

Next we evaluate $f(x)$ for those values and look for the first one that is bigger than $10^6$

```python
for i, y_val in enumerate(y_vals):
    if y_val > 10**6:
        print(f"The first element of x_vals for which y_vals > 10^6 is: {x_vals[i]}")
        break
```

Check the result by examining all the computed $f(x)$ values:

```python
y_vals
```

## Computing one-sided infinite limits with sympy

Let us also use sympy to compute the one-sided infinite limits at $x = 2$ of the function $f$ we have used above.

```python
sy.limit(f, x, 2, dir='+')
```

```python
sy.limit(f, x, 2, dir='-')
```

# Illustrating the Sandwich Rule

```python
# Define the necessary functions
f = x

print("f(x) =")
display(f)

g = sy.sin(1/ x)

print()

print("g(x) =")
display(g)
```

Plot of the product function:

```python
sy.plot((f * g, 'f(x)g(x)', {'line_style': '--'}), (x, 'y=x', {'line_style': '--'}), (-x, 'y=-x', {'line_style': '--'}), (x, -1, 1), ylim=(-1, 1), title='Plot of f(x)g(x)', legend=True);
```

sympy es capaz de calcular correctamente este límite a pesar de que el límite de g(x) no existe.

```python
sy.limit(f * g, x, 0)
```
