# Newton's Method for $\sqrt{2}$

We start by defining the function whose root we want to find:

```python
def f(x):
  return(x**2 - 2)
```

For this first and simple example we manually define the derivative of the function:

```python
def df(x):
  return(2 * x)
```

Next we define a function that takes an approximate root value and computes a single *Newton step* to obtaine the next approximation of the root.

```python
def newton_step(x):
  return(x - f(x) / df(x))
```

And we apply it to our initial guess $$\sqrt{2}\approx 1.4$$

```python
newton_step(1.4)
```

And then we repeat with our new estimate:

```python
newton_step(newton_step(1.4))
```

And again

```python
newton_step(newton_step(newton_step(newton_step(1.4))))
```

Now we use numpy, one of the most important scientific computing libraries in Python, to get a very precise estimate of the value of $\sqrt{2}$:

```python
import numpy as np
np.sqrt(2)
```

## Using a loop to make this better

```python
x0 = 1.4
print(f"Root candidate {x0}, function value {f(x0)} ")
k = 5
for i in np.arange(k):
  x0 = newton_step(x0)
  print(f"Root candidate {x0}, function value: {f(x0)} ")
```

# Newton's Method using SymPy for Function and Derivative

Sympy is a Python library for symbolic computation. That means that sympy understands that $x$ is a symbol for a variable, that $\sqrt{2}$ is a number whose square is *exactly* 2. In contrast, numpy only ever works with approximate *numeric* values. You will meet many examples of *symbolic vs numeric* when using Mathematics.

We will use sympy to solve the equation:
$$\sin(x/4) = 1$$
Think about this equation before proceeding.


```python
import sympy as sy

# Define a symbolic variable
x_sym = sy.Symbol('x')

# Define the symbolic function: sin(x)^2 - 1/2
f_sym = sy.sin(x_sym/4)**2 - sy.S(1)/2

# Compute the symbolic derivative
df_sym = sy.diff(f_sym, x_sym)

print(f"Symbolic function: {f_sym}")
print(f"Symbolic derivative: {df_sym}")
```

As you can see above, sympy knows about common elementary functions (trigonometric, exponential, etc.) and about derivation rules (the derivative of sin is cos and so on...) For this example, all we care about is getting the derivative via sympy. Now we can convert both the function and its derivative to numeric and repeat the steps we saw avove.

```python
# Convert symbolic functions to numerical functions using lambdify
f_numeric = sy.lambdify(x_sym, f_sym, 'numpy')
df_numeric = sy.lambdify(x_sym, df_sym, 'numpy')
```

```python

# Define the newton_step using the numeric functions from sympy
def newton_step_sympy(x_val):
  return x_val - f_numeric(x_val) / df_numeric(x_val)

# Perform the Newton's method computation
x0_sympy = 1.4
k_sympy = 15

print(f"Starting Newton's method with x0 = {x0_sympy}, k = {k_sympy} iterations (using sympy derived functions)\n")
print(f"Root candidate {x0_sympy}, function value {f_numeric(x0_sympy)}")
for i in np.arange(k_sympy):
  x0_sympy = newton_step_sympy(x0_sympy)
  print(f"Root candidate {x0_sympy}, function value: {f_numeric(x0_sympy)}")
```
